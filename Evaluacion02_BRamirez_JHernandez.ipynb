### Problema 9.18 de Griffiths: interfaz aire/diamante

___

### 1. Enunciado
El índice de refracción del diamante es $n_2 = 2.42$. Construir el gráfico
análogo a la Fig. 9.16 para la interfaz aire/diamante (asumir
$\mu_1 = \mu_2 = \mu_0$). En particular, calcular:

**(a)** las amplitudes a incidencia normal

**(b)** el ángulo de Brewster

**(c)** el ángulo de "cruce", en el cual las amplitudes de la onda reflejada y transmitida son iguales.

---


#### Descripción del Problema ####

Para el desarrollo de este problema se considera explícitamente el caso de polarización paralela al plano de incidencia (polarización $p$ o modos TM), donde el vector de campo eléctrico $\vec{E}$ de la onda incidente oscila dentro del plano definido por la normal a la interfaz y el vector de onda. Mediante el uso de las condiciones de frontera de Maxwell y las ecuaciones de Fresnel analíticas, se estudia la dependencia de las amplitudes reflejada y transmitida respecto al ángulo de incidencia $\theta_I$ en la interfaz aire-diamante ($n_1 = 1.0$, $n_2 = 2.42$), analizando fenómenos característicos de esta polarización como el ángulo de Brewster y el punto de cruce entre amplitudes.

___

## 2. Teoría (Griffiths, sec. 9.3)

### Abreviaturas $\alpha$ y $\beta$ (ecs. 9.106–9.108)

En el Griffiths se tienen las fórmulas de Fresnel
$$\alpha \equiv \frac{\cos\theta_T}{\cos\theta_I}, \qquad
\beta \equiv \frac{\mu_1 n_2}{\mu_2 n_1} \xrightarrow{\mu_1 = \mu_2}
\frac{n_2}{n_1}.$$

con $n_1 = 1$, $n_2 = 2.42$: $\beta = 2.42$ fijo.

A lo largo del barrido
$\theta_I \in [0, \pi/2)$, $\alpha$ crece monótonamente de $1$ a $\infty$.

### Polarización paralela (p) — ec. 9.109

$$r_p = \frac{\alpha - \beta}{\alpha + \beta}, \qquad
t_p = \frac{2}{\alpha + \beta}.$$

A incidencia normal ($\alpha = 1$):
$$r_p(0) = \frac{1 - n_2/n_1}{1 + n_2/n_1} = \frac{n_1 - n_2}{n_1 + n_2}
= -0.4152,$$
$$t_p(0) = \frac{2}{1 + n_2/n_1} = \frac{2n_1}{n_1 + n_2} = +0.5848.$$

El signo negativo de $r_p(0)$ refleja el desfase de $\pi$ de la onda
reflejada cuando $n_2 > n_1$, consistente con la ec. 9.85 para incidencia
normal.

### Polarización perpendicular (s) — problema 9.17

Griffiths trata este caso (9.3.2) con notación explícita. Forma en las
abreviaturas $\alpha, \beta$, (es lo que pide el ejercicio 9.17)
$$r_s = \frac{1 - \alpha\beta}{1 + \alpha\beta}, \qquad
t_s = \frac{2}{1 + \alpha\beta}.$$

A incidencia normal también dan $r_s(0) = (n_1 - n_2)/(n_1 + n_2) =
-0.4152$ y $t_s(0) = +0.5848$. Ambas polarizaciones coinciden en signo a incidencia normal, porque a $\theta = 0$ no hay plano de incidencia bien definido y los campos reflejados tienen el mismo desfase de $\pi$.

### Coeficientes de energía (ecs. 9.115 y 9.116)

$$R = |r|^2, \qquad
T = \frac{n_2 \cos\theta_T}{n_1 \cos\theta_I}\,|t|^2.$$

Con $\mu_1 = \mu_2$ se conserva $R + T = 1$ para ambas polarizaciones y todo
$\theta_I \in [0, \pi/2)$. Aquí no hay ángulo crítico, la reflexión total
interna requeriría $n_1 > n_2$, y en este problema $n_1 < n_2$.

### Ángulo de Brewster (ecs. 9.111 y 9.112)

En polarización paralela, $r_p = 0$ cuando $\alpha = \beta$, equivalente
geométricamente a $\theta_I + \theta_T = \pi/2$.

$$\sin^2\theta_B = \frac{1 - \beta^2}{(n_1/n_2)^2 - \beta^2},$$

con $\mu_1 = \mu_2$ se reduce a la forma simple (ec. 9.112)

$$\tan\theta_B = \frac{n_2}{n_1}.$$

con $n_2/n_1 = 2.42$ resulta $\theta_B = 67.55°$.

En polarización perpendicular **no existe** ángulo de Brewster: $r_s$ solo
se anularía si $\alpha\beta = 1$, imposible con $\beta > 1$ y $\alpha \ge 1$.
En el límite rasante $\theta_I \to \pi/2$, $r_s \to -1$.

###  Ángulo de cruce

El enunciado pide el ángulo donde "las amplitudes de la onda reflejada y
transmitida son iguales", es decir $|r| = t$ con $r, t$ reales
(no hay ángulo crítico en esta configuración).

**Paralela:** Igualando $|(\alpha - \beta)/(\alpha + \beta)| = 2/(\alpha + \beta)$:
$$|\alpha - \beta| = 2.$$
Como $\alpha$ crece de $1$ a $\infty$ y $\beta = 2.42$:

- Por debajo de Brewster ($\alpha < \beta$): $\beta - \alpha = 2 \Rightarrow
  \alpha = \beta - 2 = 0.42 < 1$, **sin solución**.
- Por encima de Brewster ($\alpha > \beta$): $\alpha - \beta = 2 \Rightarrow
  \alpha = \beta + 2 = 4.42$, **única solución**.

Despejando $\theta_I$ de $\alpha = \cos\theta_T/\cos\theta_I$ con Snell,
$$\sin^2\theta = \frac{\alpha^2 - 1}{\alpha^2 - (n_1/n_2)^2}.$$
Con $\alpha = 4.42$: $\theta_\text{cross}^{\parallel} \approx 78.06°$, y allí
$r_p = t_p = +0.2924$.

**Perpendicular:** Igualando $|(1 - \alpha\beta)/(1 + \alpha\beta)| = 2/(1 + \alpha\beta)$:
$$|1 - \alpha\beta| = 2.$$
Como $\alpha\beta \ge \beta = 2.42 > 1$, tenemos $\alpha\beta - 1 = 2$, es
decir $\alpha\beta = 3$.

Despejando, se obtiene
$$\sin^2\theta = \frac{9 n_1^2 - n_2^2}{8 n_1^2}.$$
Con $n_1 = 1$, $n_2 = 2.42$: $\theta_\text{cross}^{\perp} \approx 38.81°$, y
allí $r_s = -0.5$, $t_s = +0.5$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm
from ipywidgets import interact, FloatSlider, Dropdown

# Configuración de matplotlib
plt.rcParams.update({
    "figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.25,
    "font.size": 10, "axes.titlesize": 11, "lines.linewidth": 2,
})

# Interfaz aire / diamante (Problema 9.18) ---
N1, N2 = 1.00, 2.42       # índices de refracción
BETA    = N2 / N1         # β ≡ n₂/n₁  (ec. 9.108, con μ₁ = μ₂)

# Azul = perpendicular (s), rojo = paralela (p)
COLOR_S, COLOR_P = "#1f5fa8", "#c0392b"

print(f"n1 = {N1}  (aire)   n2 = {N2}  (diamante)")
```

```python
def theta_transmitted(theta_I, n1=N1, n2=N2):
    """Ángulo transmitido por la ley de Snell (ec. 9.100)."""
    return np.arcsin((n1 / n2) * np.sin(theta_I))


def fresnel(theta_I, n1=N1, n2=N2):
    """
    Coeficientes de amplitud de Fresnel en la forma α–β de Griffiths.

    Devuelve (r_perp, t_perp, r_par, t_par) para cada θ_I (radianes).
    Todo vectorizado; fórmulas analíticas cerradas, sin optimización.

    Paralela (ec. 9.109)             Perpendicular (problema 9.17)
        r_par  = (α − β)/(α + β)         r_perp = (1 − αβ)/(1 + αβ)
        t_par  = 2/(α + β)               t_perp = 2/(1 + αβ)

    donde α ≡ cos θ_T / cos θ_I  y  β ≡ n₂/n₁  (ecs. 9.106–9.108, μ₁ = μ₂).
    Las formas con n₁, n₂ salen de multiplicar numerador y denominador
    por n₁ cos θ_I (paralela) o n₁ cos θ_I · n₁ cos θ_T (perpendicular).
    """
    theta_I = np.asarray(theta_I, dtype=float)
    cI = np.cos(theta_I)
    cT = np.cos(theta_transmitted(theta_I, n1, n2))

    # Paralela — ec. 9.109, forma α − β / α + β
    denom_par = n1 * cT + n2 * cI
    r_par = (n1 * cT - n2 * cI) / denom_par      # ← signo Griffiths
    t_par = (2 * n1 * cI)       / denom_par

    # Perpendicular — problema 9.17, forma 1 − αβ / 1 + αβ
    denom_perp = n1 * cI + n2 * cT
    r_perp = (n1 * cI - n2 * cT) / denom_perp
    t_perp = (2 * n1 * cI)       / denom_perp

    return r_perp, t_perp, r_par, t_par


def fresnel_RT(theta_I, n1=N1, n2=N2):
    """
    Coeficientes de energía R y T (ecs. 9.115 y 9.116).

    Devuelve (R_perp, T_perp, R_par, T_par) para cada θ_I (radianes).
        R = |r|²,   T = (n₂ cos θ_T)/(n₁ cos θ_I) · |t|².
    """
    theta_I = np.asarray(theta_I, dtype=float)
    thT = theta_transmitted(theta_I, n1, n2)
    r_perp, t_perp, r_par, t_par = fresnel(theta_I, n1, n2)

    factor = (n2 * np.cos(thT)) / (n1 * np.cos(theta_I))
    return (r_perp**2, factor * t_perp**2,
            r_par**2,  factor * t_par**2)


def theta_brewster(n1=N1, n2=N2):
    """Ángulo de Brewster (ec. 9.112): tan θ_B = n₂/n₁."""
    return np.arctan(n2 / n1)


def theta_crossover_perp(n1=N1, n2=N2):
    """
    Ángulo de cruce, polarización perpendicular (problema 9.17).

    Sale de |r_perp| = t_perp  ⟺  αβ = 3  (con β = n₂/n₁ > 1, la única
    rama posible es αβ − 1 = 2).  Despejando:
        sin²θ = (9 n₁² − n₂²) / (8 n₁²).
    """
    s2 = (9 * n1**2 - n2**2) / (8 * n1**2)
    return np.arcsin(np.sqrt(s2)) if 0 < s2 <= 1 else np.nan


def theta_crossover_par(n1=N1, n2=N2):
    """
    Ángulo de cruce, polarización paralela (derivación propia).

    Sale de |r_par| = t_par  ⟺  |α − β| = 2.  Como α ≥ 1 y β = n₂/n₁:
      · rama α = β − 2 < 1 no es física (α ≥ 1 siempre),
      · única solución: α = β + 2.
    Despejando:
        sin²θ = (α² − 1) / (α² − (n₁/n₂)²).
    Con los valores del problema, α = 2.42 + 2 = 4.42 y θ ≈ 78.06°.
    """
    alpha_x = (n2 / n1) + 2.0
    s2 = (alpha_x**2 - 1) / (alpha_x**2 - (n1 / n2)**2)
    return np.arcsin(np.sqrt(s2)) if 0 < s2 <= 1 else np.nan
```

```python
print("=" * 64)
print("PROBLEMA 9.18 — Interfaz aire / diamante  (n₁ = 1, n₂ = 2.42)")
print("=" * 64)

# ---------------------------------------------------------------
# (a) Incidencia normal
# ---------------------------------------------------------------
r_perp_0, t_perp_0, r_par_0, t_par_0 = fresnel(0.0)

print("\n(a) Amplitudes a incidencia normal (θ_I = 0°):")
print(f"    r_perp = {r_perp_0:+.4f}      t_perp = {t_perp_0:+.4f}")
print(f"    r_par  = {r_par_0:+.4f}      t_par  = {t_par_0:+.4f}")
print(f"    → ambas polarizaciones dan r = (n₁−n₂)/(n₁+n₂) = "
      f"{r_perp_0:+.4f} (desfase π)")
print(f"      y t = 2n₁/(n₁+n₂) = {t_perp_0:+.4f}")

# ---------------------------------------------------------------
# (b) Ángulo de Brewster
# ---------------------------------------------------------------
thB = theta_brewster()
thT_at_B = theta_transmitted(thB)
r_par_B, t_par_B = fresnel(thB)[2:]
sum_at_B = np.degrees(thB) + np.degrees(thT_at_B)

print(f"\n(b) Ángulo de Brewster:")
print(f"    β = n₂/n₁ = {BETA:.4f}  →  θ_B = arctan(β) = "
      f"{np.degrees(thB):.2f}°")
print(f"    verificación: r_par(θ_B) = {r_par_B:+.2e}")
print(f"                  θ_I + θ_T = {sum_at_B:.6f}°  (debe ser 90°)")

# ---------------------------------------------------------------
# (c) Ángulo de cruce — paralela (inciso literal del 9.18)
# ---------------------------------------------------------------
thx_par = theta_crossover_par()
r_par_x, t_par_x = fresnel(thx_par)[2:]

print(f"\n(c) Ángulo de cruce, polarización paralela (r_par = t_par):")
print(f"    α = β + 2 = {BETA + 2:.4f}  →  θ ≈ {np.degrees(thx_par):.2f}°")
print(f"    r_par = {r_par_x:+.4f},  t_par = {t_par_x:+.4f}")

# ---------------------------------------------------------------
# (d) Extensión — crossover perpendicular (problema 9.17)
# ---------------------------------------------------------------
thx_perp = theta_crossover_perp()
r_perp_x, t_perp_x = fresnel(thx_perp)[:2]

print(f"\n[Extensión] Cruce perpendicular (|r_perp| = t_perp):")
print(f"    αβ = 3  →  θ ≈ {np.degrees(thx_perp):.2f}°")
print(f"    r_perp = {r_perp_x:+.4f},  t_perp = {t_perp_x:+.4f}")

# ===============================================================
# Asserts automáticos
# ===============================================================
# (a) Incidencia normal — signo correcto en AMBAS polarizaciones
assert np.isclose(r_perp_0, -0.4152, atol=1e-4), r_perp_0
assert np.isclose(r_par_0,  -0.4152, atol=1e-4), r_par_0
assert np.isclose(t_perp_0, +0.5848, atol=1e-4), t_perp_0
assert np.isclose(t_par_0,  +0.5848, atol=1e-4), t_par_0

# (b) Brewster: r_par = 0 y θ_I + θ_T = 90°
assert abs(r_par_B) < 1e-12, r_par_B
assert np.isclose(sum_at_B, 90.0, atol=1e-10), sum_at_B
assert np.isclose(np.degrees(thB), 67.55, atol=0.01)

# (c) Crossover paralela: r_par = t_par, θ ≈ 78.06°
assert np.isclose(r_par_x, t_par_x, atol=1e-12)
assert np.isclose(np.degrees(thx_par), 78.06, atol=0.01)

# (d) Crossover perpendicular: |r_perp| = t_perp, θ ≈ 38.81°
assert np.isclose(abs(r_perp_x), t_perp_x, atol=1e-12)
assert np.isclose(np.degrees(thx_perp), 38.81, atol=0.01)

print("\n✓ Todos los asserts pasaron.")
```

```python
def boundary_conditions(theta_I, n1=N1, n2=N2):
    """
    Residuos de las condiciones de frontera en z = 0 (Griffiths §9.3.1).

    Con μ₁ = μ₂ = μ₀ y sin cargas ni corrientes superficiales, la física
    exige que las componentes tangenciales de E y B sean continuas.
    En términos de las amplitudes y θ_T (por Snell):

        s:  1 + r_perp = t_perp
            n₁ cosθ_I (1 − r_perp) = n₂ cosθ_T · t_perp

        p:  cosθ_I (1 + r_par) = cosθ_T · t_par
            n₁ (1 − r_par) = n₂ · t_par

    Devuelve (res_s_E, res_s_B, res_p_E, res_p_B), cada uno con la forma
    LHS − RHS, para cada θ_I del arreglo de entrada.
    """
    theta_I = np.asarray(theta_I, dtype=float)
    cI = np.cos(theta_I)
    cT = np.cos(theta_transmitted(theta_I, n1, n2))
    r_perp, t_perp, r_par, t_par = fresnel(theta_I, n1, n2)

    # s-polarización: E_y continua, B_x continua
    res_s_E = (1 + r_perp) - t_perp
    res_s_B = n1 * cI * (1 - r_perp) - n2 * cT * t_perp

    # p-polarización: E_x continua, B_y continua
    res_p_E = cI * (1 + r_par) - cT * t_par
    res_p_B = n1 * (1 - r_par) - n2 * t_par

    return res_s_E, res_s_B, res_p_E, res_p_B


# ============================================================
# Barrido sobre todo el rango de incidencia
# ============================================================
residuos = boundary_conditions(theta_rad)
etiquetas = [
    r"s:  $E_y$ continua",
    r"s:  $B_x$ continua",
    r"p:  $E_x$ continua",
    r"p:  $B_y$ continua",
]

print("Residuos de las condiciones de frontera en z = 0")
print("(todos deben ser del orden de la precisión de máquina)")
print("─" * 60)
for lab, r in zip(etiquetas, residuos):
    print(f"  {lab:<26}  máx|residuo| = {np.max(np.abs(r)):.2e}")

# ------------------------------------------------------------
# Diagnóstico puntual: veamos los campos a 0°, 45° y 80°
# ------------------------------------------------------------
print()
print("Residuos a ángulos seleccionados:")
print("─" * 60)
print(f"  {'θ_I':>5}°   " + "   ".join(f"{lab:>22}" for lab in etiquetas))
for th_deg in [0, 45, 80]:
    th = np.deg2rad(th_deg)
    vals = boundary_conditions(th)
    row = "   ".join(f"{float(v):+22.2e}" for v in vals)
    print(f"  {th_deg:>5}°   {row}")

# ------------------------------------------------------------
# Assert global
# ------------------------------------------------------------
max_res = max(np.max(np.abs(r)) for r in residuos)
assert max_res < 1e-14, max_res

print()
print(f"✓ Las cuatro condiciones de frontera se satisfacen "
      f"a precisión de máquina (máx|residuo| ≈ {max_res:.2e}).")
print("  Esto verifica la implementación de r y t por un camino "
      "de código distinto al de las fórmulas α–β.")
```

```python
def snapshot_EM(theta_I_deg, phi, pol):
    """
    Snapshot del campo transverso en el plano (x, z).

    · Polarización s: grafica E_y (tangencial, continuo en z=0).
    · Polarización p: grafica c·B_y (tangencial, continuo en z=0).
      Se elige B_y en lugar de E_x porque E_x lleva un factor cos θ_I
      que apaga la amplitud a ángulos altos, mientras que B_y es el
      análogo natural a E_y de la polarización s.

    φ ≡ ωt parametriza el instante.
    """
    thI = np.deg2rad(theta_I_deg)
    thT = theta_transmitted(thI, N1, N2)
    r_perp, t_perp, r_par, t_par = fresnel(thI, N1, N2)

    k1 = 2 * np.pi
    k2 = (N2 / N1) * k1

    x = np.linspace(-2.5, 2.5, 400)
    z = np.linspace(-2.5, 2.5, 400)
    X, Z = np.meshgrid(x, z)

    phi_I = k1 * (np.sin(thI) * X + np.cos(thI) * Z) - phi
    phi_R = k1 * (np.sin(thI) * X - np.cos(thI) * Z) - phi
    phi_T = k2 * (np.sin(thT) * X + np.cos(thT) * Z) - phi

    if pol == "s":
        # E_y (componente tangencial de E)
        E_I = np.where(Z < 0, np.cos(phi_I), 0.0)
        E_R = np.where(Z < 0, r_perp * np.cos(phi_R), 0.0)
        E_T = np.where(Z > 0, t_perp * np.cos(phi_T), 0.0)
        field = E_I + E_R + E_T
        titulo = r"$\mathrm{Re}[E_y]$  (polarización s)"
    else:
        # B_y (componente tangencial de B).  Con μ₁=μ₂ y c=1:
        #   BC: n₁(1 − r_par) = n₂ t_par  ⟹  B_y continua en z = 0.
        B_I = np.where(Z < 0,  N1 * np.cos(phi_I), 0.0)
        B_R = np.where(Z < 0, -N1 * r_par * np.cos(phi_R), 0.0)
        B_T = np.where(Z > 0,  N2 * t_par * np.cos(phi_T), 0.0)
        field = B_I + B_R + B_T
        titulo = r"$\mathrm{Re}[c\,B_y]$  (polarización p)"

    fig, ax = plt.subplots(figsize=(6.4, 5.4))
    vmax = max(1.0, np.abs(field).max())
    im = ax.pcolormesh(X, Z, field, cmap="RdBu_r",
                       norm=TwoSlopeNorm(vcenter=0, vmin=-vmax, vmax=vmax),
                       shading="auto")
    ax.axhline(0, color="k", lw=1.2)
    ax.set(xlabel="x (u.a.)", ylabel="z (u.a.)", aspect="equal",
           title=f"{titulo}\n"
                 f"$\\theta_I$ = {theta_I_deg:.0f}°   $\\varphi = \\omega t$ = {phi:.2f}")
    fig.colorbar(im, ax=ax, shrink=0.85, label="amplitud")
    plt.tight_layout()
    plt.show()


interact(
    snapshot_EM,
    theta_I_deg=FloatSlider(value=45, min=0, max=89, step=1,
                            description="θ_I (°)", continuous_update=False),
    phi=FloatSlider(value=0.0, min=0.0, max=6.283, step=0.05,
                    description="ωt", continuous_update=False),
    pol=Dropdown(options=["s", "p"], value="s", description="pol."),
)
```

## 3. Interpretación física

### 3.1 ¿Por qué no hay reflexión total interna en aire/diamante?

Porque para $n_1 < n_2$ el segundo medio es ópticamente más denso, así que la
ley de Snell
$$\sin\theta_T = \frac{n_1}{n_2}\sin\theta_I$$
admite solución real $\theta_T$ para todo $\theta_I \in [0, \pi/2]$. El ángulo
crítico $\theta_c = \arcsin(n_2/n_1)$ no existe porque es mayor
que 1. Podría existir reflexión total interna si el sentido de propagación fuera el inverso (diamante → aire).

### 3.2 Amplitudes a incidencia normal

En incidencia normal $\theta_I = \theta_T = 0$, las fórmulas de Fresnel
colapsan a una única expresión para ambas polarizaciones:
$$r(0) = \frac{n_1 - n_2}{n_1 + n_2} = \frac{1 - 2.42}{1 + 2.42} = -0.4152,$$
$$t(0) = \frac{2 n_1}{n_1 + n_2} = +0.5848.$$

El signo negativo de $r(0)$ es físico, no un artefacto de convención. Cuando
$n_2 > n_1$ la onda reflejada sale con un desfase de $\pi$ respecto a la
incidente. Ese desfase se observa igual para las dos polarizaciones, porque a
incidencia normal no hay plano de incidencia bien definido y ambas elecciones
de "dirección positiva" coinciden.

### 3.3 Ángulo de Brewster

En polarización paralela, cuando $\theta_I = \theta_B$ el numerador de $r_\parallel$ se anula. Físicamente, la onda reflejada en una interfaz dieléctrica es la superposición de la radiación emitida por los dipolos inducidos en el medio 2. Esos dipolos oscilan en la dirección del campo $\mathbf{E}$ transmitido. Cuando $\theta_I + \theta_T = 90°$, la dirección de oscilación de los dipolos coincide con la dirección de propagación del rayo reflejado, y un dipolo no radia en su propia dirección de oscilación, entonces la onda reflejada desaparece.

En polarización perpendicular los dipolos oscilan siempre perpendicular al
plano de incidencia, así que nunca se alinean con la dirección reflejada. Por lo tanto, $r_\perp$ no se anula para ningún ángulo físico y en el límite tiende a $-1$.

### 3.4 Ángulo de cruce

El enunciado pide el ángulo donde las amplitudes reflejada y transmitida
tienen igual magnitud. Como en este problema no hay desfases complejos
(no hay ángulo crítico), $r$ y $t$ son reales y la condición es $|r| = t$.

**Paralela:**
$$|r_\parallel| = t_\parallel \;\Longleftrightarrow\;
\frac{|\alpha - \beta|}{\alpha + \beta} = \frac{2}{\alpha + \beta}
\;\Longleftrightarrow\; |\alpha - \beta| = 2.$$

Como $\alpha$ crece de $1$ a $\infty$ y $\beta = 2.42 > 2$:
- debajo de Brewster ($\alpha < \beta$): $\alpha = \beta - 2 = 0.42 < 1$ (fuera del rango físico de $\alpha$).
- encima de Brewster ($\alpha > \beta$): $\alpha = \beta + 2 = 4.42$ (única solución válida).

Despejando $\theta_I$ de $\alpha = \cos\theta_T/\cos\theta_I$ junto con Snell,
$$\sin^2\theta = \frac{\alpha^2 - 1}{\alpha^2 - (n_1/n_2)^2}.$$
Numéricamente $\theta_\text{cross}^\parallel \approx 78.06°$, y allí
$r_\parallel = t_\parallel = +0.2924$. Se obtiene el valor para el inciso (c) del problema.

**Perpendicular (extensión):**
$$|r_\perp| = t_\perp \;\Longleftrightarrow\; \alpha\beta = 3,$$
de donde
$$\sin^2\theta = \frac{9 n_1^2 - n_2^2}{8 n_1^2}, \qquad
\theta_\text{cross}^\perp \approx 38.81°,$$
con $r_\perp = -0.5$, $t_\perp = +0.5$.

### 3.5 Validación desde las condiciones de frontera

Las amplitudes de Fresnel se derivan de las condiciones de frontera en
$z = 0$. Continuidad de las componentes tangenciales de
$\mathbf{E}$ y $\mathbf{B}$ cuando $\mu_1 = \mu_2$ y no hay cargas ni
corrientes superficiales. Por lo tanto, se busca verificar dicha continuidad de
forma numérica e independiente del código que evalúa las fórmulas $\alpha$–$\beta$:

- **s:** $E_y$ continua y $B_x$ continua.
- **p:** $E_x$ continua y $B_y$ continua.

Los cuatro residuos resultan del orden de la precisión de máquina
($\sim 10^{-16}$) sobre todo el barrido angular. Esto constituye una
verificación genuina de la implementación de $r$ y $t$, porque usa un camino
de código distinto al de las fórmulas de Fresnel.


### 3.6 La visualización interactiva: por qué se grafica $c\,B_y$ en polarización p

En polarización s el campo eléctrico tiene una sola componente, $E_y$,
tangencial a la interfaz. En polarización p, $\mathbf{E}$ está contenido en el plano de incidencia y sus componentes $E_x$, $E_z$ son tangencial y normal respectivamente. La componente tangencial es $E_x$, pero lleva un factor $\cos\theta_I$ que apaga la amplitud a ángulos altos, lo que dificulta ver el patrón de interferencia en el medio 1.

La alternativa es $c\,B_y$, y para p $\mathbf{B}$ es perpendicular al
plano de incidencia, así que $B_y$ es la componente tangencial, es continua
en $z = 0$ y no lleva el factor $\cos\theta_I$. Es el análogo exacto de
$E_y$ para la polarización perpendicular.


## 4. Retrospectiva: qué se entendió mejor visualizando

La aparición y desaparición del ángulo de
Brewster según la polarización.

### ¿Qué muestran las ecuaciones?

Las dos fórmulas de Fresnel, escritas en la forma α–β de Griffiths, son casi idénticas
$$r_\parallel = \frac{\alpha - \beta}{\alpha + \beta}, \qquad
r_\perp = \frac{1 - \alpha\beta}{1 + \alpha\beta}.$$

la única diferencia está en cómo entra el producto $\alpha\beta$. En la
paralela aparece como una **resta** en el numerador, en la perpendicular
como una **resta de $1$** contra el producto.

### ¿Qué mejora la visualización?

En el widget de la sección 3.6, moviendo el slider de $\theta_I$ con
polarización p seleccionada, se observan tres cosas:

1. En el medio 1 ($z < 0$) hay un patrón de interferencia entre la onda
   incidente y la reflejada, salvo en un ángulo preciso ($\theta_B$).
2. Exactamente en $\theta_B$ ese patrón desaparece, se ve una sola onda
   propagándose hacia la interfaz, y a onda reflejada se anuló.
3. Cambiando a polarización s, ese ángulo no existe, la onda reflejada
   nunca desaparece, aunque que se recorra todo el rango de incidencia.

### ¿Por qué ocurre físicamente?

La onda reflejada no es un fenómeno cinemático, es la superposición de la radiación emitida por los dipolos inducidos en el medio 2. Los electrones del dieléctrico oscilan en la dirección del campo eléctrico transmitido. Un dipolo no radia en su propia dirección de oscilación.

En polarización paralela, cuando $\theta_I + \theta_T = 90°$, la dirección
de oscilación de los dipolos coincide exactamente con la dirección de
propagación del rayo reflejado. Los dipolos no pueden radiar hacia atrás y
la onda reflejada desaparece. Esa condición $\theta_I + \theta_T = 90°$ es
algebraicamente equivalente a $\alpha = \beta$, y por eso $r_\parallel = 0$
en ese ángulo preciso.

En polarización perpendicular, los dipolos oscilan siempre perpendiculares
al plano de incidencia, así que jamás se alinean con la dirección reflejada, no hay Brewster. En el límite rasante $\theta_I \to \pi/2$, la onda reflejada
existe con amplitud máxima ($r_\perp \to -1$), no con amplitud nula.

### ¿Por qué la visualización ayudó más que solo las ecuaciones?

La fórmula $r_\parallel = (\alpha-\beta)/(\alpha+\beta)$ permite ver hay un cero en $\alpha = \beta$, pero no permite ver por qué ese cero debería ser físicamente relevante. Lo que hace la visualización es convertir ese cero en un evento visible (la onda reflejada se apaga) y con eso se cae a la pregunta "¿qué representa ese ángulo?". Acá se tiene la interpretación
dipolar, que es la que finalmente explica el fenómeno. En otras palabras,
la fórmula da el número, la visualización da la pregunta, y la respuesta a
esa pregunta se da fisicamente.

Lo que no se entendía antes de ver el widget era que la
condición $\theta_I + \theta_T = 90°$ no es una coincidencia algebraica
sino una condición geométrica que alinea la oscilación de los dipolos con
la dirección reflejada. Ver la onda reflejada desaparecer mientras el rayo
transmitido sigue propagándose hace evidente que la anulación no es un
cálculo sino una imposibilidad física de radiar en esa dirección.
